```python
df["MPG"] = df["MIN"] / df["GP"]

df["PPG"] = df["PTS"] / df["GP"]
df["APG"] = df["AST"] / df["GP"]
df["RPG"] = df["REB"] / df["GP"]
df["SPG"] = df["STL"] / df["GP"]
df["BPG"] = df["BLK"] / df["GP"]
df["TOPG"] = df["TOV"] / df["GP"]

df["TS_PCT"] = df["PTS"] / (
    2 * (df["FGA"] + 0.44 * df["FTA"])
)

df["EFG_PCT"] = (
    df["FGM"] + 0.5 * df["FG3M"]
) / df["FGA"]

df["THREE_PA_RATE"] = df["FG3A"] / df["FGA"]

df["FTA_RATE"] = df["FTA"] / df["FGA"]
```

```python
# Kobe career reference values
total_gp = df["GP"].sum()

kobe_reference = {
    "MPG": df["MIN"].sum() / total_gp,
    "PPG": df["PTS"].sum() / total_gp,
    "APG": df["AST"].sum() / total_gp,
    "RPG": df["REB"].sum() / total_gp,
    "SPG": df["STL"].sum() / total_gp,
    "BPG": df["BLK"].sum() / total_gp,
    "TOPG": df["TOV"].sum() / total_gp,

    "FG_PCT": df["FGM"].sum() / df["FGA"].sum(),
    "FG3_PCT": df["FG3M"].sum() / df["FG3A"].sum(),
    "FT_PCT": df["FTM"].sum() / df["FTA"].sum(),

    "TS_PCT": df["PTS"].sum() / (
        2 * (df["FGA"].sum() + 0.44 * df["FTA"].sum())
    ),

    "EFG_PCT": (
        df["FGM"].sum() + 0.5 * df["FG3M"].sum()
    ) / df["FGA"].sum(),

    "THREE_PA_RATE": df["FG3A"].sum() / df["FGA"].sum(),
    "FTA_RATE": df["FTA"].sum() / df["FGA"].sum()
}

kobe_reference
```

```python
def similarity_score(reference, candidate):
    if reference == 0:
        return 100 if candidate == 0 else 0

    difference = abs(candidate - reference)
    percentage_difference = difference / abs(reference)

    return max(0, 100 * (1 - percentage_difference))
```

```python
from nba_api.stats.endpoints import leaguedashplayerstats
import pandas as pd
import time

kobe_advanced_seasons = []

for season in df["SEASON_ID"]:
    try:
        season_stats = leaguedashplayerstats.LeagueDashPlayerStats(
            season=season,
            season_type_all_star="Regular Season",
            measure_type_detailed_defense="Advanced",
            per_mode_detailed="PerGame"
        ).get_data_frames()[0]

        kobe_season = season_stats[
            season_stats["PLAYER_ID"] == 977
        ].copy()

        if not kobe_season.empty:
            kobe_season["SEASON_ID"] = season
            kobe_advanced_seasons.append(kobe_season)

        print(season, "done")

        time.sleep(0.5)

    except Exception as e:
        print(season, "ERROR:", e)

kobe_advanced = pd.concat(
    kobe_advanced_seasons,
    ignore_index=True
)

print("\nKobe advanced seasons:", kobe_advanced.shape)
```

```python
key_advanced = [
    "OFF_RATING",
    "DEF_RATING",
    "NET_RATING",
    "AST_PCT",
    "AST_TO",
    "REB_PCT",
    "E_TOV_PCT",
    "USG_PCT",
    "TS_PCT",
    "EFG_PCT",
    "PACE",
    "PIE"
]

coverage = pd.DataFrame({
    "metric": key_advanced,
    "seasons_available": [
        kobe_advanced[m].notna().sum()
        for m in key_advanced
    ],
    "seasons_missing": [
        kobe_advanced[m].isna().sum()
        for m in key_advanced
    ]
})

coverage
```

```python
advanced_reference_metrics = [
    "OFF_RATING",
    "DEF_RATING",
    "NET_RATING",
    "AST_PCT",
    "AST_TO",
    "REB_PCT",
    "E_TOV_PCT",
    "USG_PCT",
    "TS_PCT",
    "EFG_PCT",
    "PACE",
    "PIE"
]

# Advanced stats are per-game/per-possession measures,
# so use total minutes as the career weighting factor.
kobe_advanced["TOTAL_MIN"] = (
    kobe_advanced["MIN"] * kobe_advanced["GP"]
)

kobe_advanced_reference = {}

for metric in advanced_reference_metrics:
    valid = kobe_advanced[["TOTAL_MIN", metric]].dropna()

    kobe_advanced_reference[metric] = (
        (valid[metric] * valid["TOTAL_MIN"]).sum()
        / valid["TOTAL_MIN"].sum()
    )

kobe_advanced_reference_df = pd.DataFrame({
    "Metric": kobe_advanced_reference.keys(),
    "Kobe_Value": kobe_advanced_reference.values()
})

kobe_advanced_reference_df.round(3)
```

```python
# Shot-location columns used for the similarity model
shot_zone_fga = [
    "Restricted Area_FGA",
    "In The Paint (Non-RA)_FGA",
    "Mid-Range_FGA",
    "Corner 3_FGA",
    "Above the Break 3_FGA",
    "Backcourt_FGA"
]

# Filter to our 70-player candidate pool
candidate_shots = test_df[
    test_df["PLAYER_ID"].isin(candidate_pool["PLAYER_ID"])
].copy()

print("Candidate players:", candidate_shots["PLAYER_ID"].nunique())
print("Rows:", candidate_shots.shape)
```

```python
# Total shots represented by the six non-overlapping zones
candidate_shots["TOTAL_ZONE_FGA"] = candidate_shots[shot_zone_fga].sum(axis=1)

# Convert each zone into a percentage of the player's shots
for col in shot_zone_fga:
    zone_name = col.replace("_FGA", "")
    candidate_shots[f"{zone_name}_FREQ"] = (
        candidate_shots[col] / candidate_shots["TOTAL_ZONE_FGA"] * 100
    )

print(
    candidate_shots[
        ["PLAYER_NAME"] +
        [col.replace("_FGA", "_FREQ") for col in shot_zone_fga]
    ].head(10).round(2)
)
```

## Day 6 — Similarity Calculations

The goal for this session is to combine the datasets collected so far and calculate similarity scores between Kobe Bryant and the 70-player comparison pool.

I'll calculate scores at both the individual-metric and category levels. Categories that still lack reliable data will remain unscored until that data is available.

```python
from pathlib import Path
import pandas as pd
import numpy as np

# Load the datasets collected so far
traditional = pd.read_csv(
    "../data/processed/candidate_traditional_stats_2025_26.csv"
)

advanced = pd.read_csv(
    "../data/processed/candidate_advanced_clean_2025_26.csv"
)

shots = pd.read_csv(
    "../data/processed/candidate_shot_profile_2025_26.csv"
)

# Keep only the advanced metrics needed for this model
advanced_columns = [
    "PLAYER_ID",
    "OFF_RATING",
    "DEF_RATING",
    "NET_RATING",
    "AST_PCT",
    "AST_TO",
    "REB_PCT",
    "E_TOV_PCT",
    "USG_PCT",
    "PIE"
]

advanced_keep = advanced[advanced_columns].copy()

# Combine everything using player IDs
candidate_data = traditional.merge(
    advanced_keep,
    on="PLAYER_ID",
    how="inner"
)

candidate_data = candidate_data.merge(
    shots.drop(columns=["PLAYER_NAME"]),
    on="PLAYER_ID",
    how="inner"
)

print("Rows:", len(candidate_data))
print("Unique players:", candidate_data["PLAYER_ID"].nunique())

candidate_data[
    ["PLAYER_NAME", "PTS", "AST", "USG_PCT", "NET_RATING"]
].head()
```

```python
kobe_raw = pd.read_csv("../data/raw/kobe_career_stats.csv")

total_gp = kobe_raw["GP"].sum()
total_fga = kobe_raw["FGA"].sum()
total_fg3a = kobe_raw["FG3A"].sum()
total_fta = kobe_raw["FTA"].sum()

kobe_traditional_reference = {
    "PTS": kobe_raw["PTS"].sum() / total_gp,
    "AST": kobe_raw["AST"].sum() / total_gp,
    "FGA": total_fga / total_gp,
    "STL": kobe_raw["STL"].sum() / total_gp,
    "BLK": kobe_raw["BLK"].sum() / total_gp,

    "FG_PCT": kobe_raw["FGM"].sum() / total_fga,

    "FG3_PCT": (
        kobe_raw["FG3M"].sum() / total_fg3a
    ),

    "FT_PCT": (
        kobe_raw["FTM"].sum() / total_fta
    ),

    "TS_PCT": (
        kobe_raw["PTS"].sum()
        / (2 * (
            total_fga + 0.44 * total_fta
        ))
    ),

    "EFG_PCT": (
        (kobe_raw["FGM"].sum()
         + 0.5 * kobe_raw["FG3M"].sum())
        / total_fga
    )
}

kobe_traditional_reference_df = pd.DataFrame(
    list(kobe_traditional_reference.items()),
    columns=["Metric", "Kobe_Value"]
)

output_path = Path(
    "../data/processed/kobe_traditional_reference.csv"
)

kobe_traditional_reference_df.to_csv(
    output_path,
    index=False
)

kobe_traditional_reference_df.round(3)
```

```python
# Load Kobe's advanced-stat reference
advanced_ref_df = pd.read_csv(
    "../data/processed/kobe_advanced_reference.csv"
)

kobe_advanced_values = dict(
    zip(
        advanced_ref_df["Metric"],
        advanced_ref_df["Kobe_Value"]
    )
)

# Use traditional career-total calculations for these two metrics
kobe_advanced_values.pop("TS_PCT", None)
kobe_advanced_values.pop("EFG_PCT", None)

# Combine traditional and advanced reference values
kobe_reference_values = {
    **kobe_traditional_reference,
    **kobe_advanced_values
}

# Add Kobe's career shooting-tendency profile
kobe_shot_ref = pd.read_csv(
    "../data/processed/kobe_shot_profile_career.csv"
).iloc[0]

shot_metrics = [
    "Restricted Area_FREQ",
    "In The Paint (Non-RA)_FREQ",
    "Mid-Range_FREQ",
    "Corner 3_FREQ",
    "Above the Break 3_FREQ",
    "Backcourt_FREQ"
]

for metric in shot_metrics:
    kobe_reference_values[metric] = kobe_shot_ref[metric]

print("Reference metrics available:", len(kobe_reference_values))
```

```python
category_metrics = {
    "Stature": [],

    "Shooting Tendencies": shot_metrics,

    "Portability": [],

    "Passing": [
        "AST",
        "AST_PCT",
        "AST_TO"
    ],

    "Usage & Creation": [
        "PTS",
        "FGA",
        "USG_PCT"
    ],

    "Defense": [
        "STL",
        "BLK",
        "DEF_RATING"
    ],

    "Athleticism": [],

    "Dominance": [
        "NET_RATING",
        "PIE"
    ],

    "Shooting Ability": [
        "FG_PCT",
        "FG3_PCT",
        "FT_PCT",
        "TS_PCT",
        "EFG_PCT"
    ],

    "Play Style": []
}

for category, metrics in category_metrics.items():
    available = [
        metric for metric in metrics
        if metric in candidate_data.columns
        and metric in kobe_reference_values
    ]

    print(f"{category}: {len(available)} metrics available")
```

```python
def percentage_similarity(reference, candidate):
    if pd.isna(reference) or pd.isna(candidate):
        return np.nan

    if np.isclose(reference, 0):
        return 100.0 if np.isclose(candidate, 0) else 0.0

    difference = abs(candidate - reference)

    percentage_difference = difference / abs(reference)

    return max(0.0, 100 * (1 - percentage_difference))


# Store the similarity score for each individual metric
metric_similarity = candidate_data[
    ["PLAYER_ID", "PLAYER_NAME"]
].copy()

# Store the average similarity score for each category
category_similarity = candidate_data[
    ["PLAYER_ID", "PLAYER_NAME"]
].copy()

for category, metrics in category_metrics.items():

    available_metrics = [
        metric for metric in metrics
        if metric in candidate_data.columns
        and metric in kobe_reference_values
    ]

    # Leave categories without reliable data unscored
    if not available_metrics:
        category_similarity[f"{category}_SIM"] = np.nan
        continue

    for metric in available_metrics:
        similarity_column = f"{metric}_SIM"

        if similarity_column not in metric_similarity.columns:
            reference = kobe_reference_values[metric]

            metric_similarity[similarity_column] = (
                candidate_data[metric].apply(
                    lambda value: percentage_similarity(
                        reference,
                        value
                    )
                )
            )

    # Average the metric scores within this category
    similarity_columns = [
        f"{metric}_SIM"
        for metric in available_metrics
    ]

    category_similarity[f"{category}_SIM"] = (
        metric_similarity[similarity_columns].mean(axis=1)
    )

print("Individual metric similarity scores:")
display(metric_similarity.head())

print("Category similarity scores:")
display(category_similarity.head())
```

```python
available_category_columns = [
    f"{category}_SIM"
    for category, metrics in category_metrics.items()
    if any(
        metric in candidate_data.columns
        and metric in kobe_reference_values
        for metric in metrics
    )
]

category_similarity["CATEGORIES_SCORED"] = (
    category_similarity[available_category_columns]
    .notna()
    .sum(axis=1)
)

category_similarity["PROVISIONAL_SCORE"] = (
    category_similarity[available_category_columns]
    .mean(axis=1)
)

provisional_ranking = category_similarity.sort_values(
    "PROVISIONAL_SCORE",
    ascending=False
)

display(
    provisional_ranking[
        [
            "PLAYER_NAME",
            "PROVISIONAL_SCORE",
            "CATEGORIES_SCORED"
        ]
    ].head(10).round(2)
)
```
