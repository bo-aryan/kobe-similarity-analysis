```python
df["MPG"] = df["MIN"] / df["GP"]

df["PPG"] = df["PTS"] / df["GP"]
df["APG"] = df["AST"] / df["GP"]
df["RPG"] = df["REB"] / df["GP"]
df["SPG"] = df["STL"] / df["GP"]
df["BPG"] = df["BLK"] / df["GP"]
df["TOPG"] = df["TOV"] / df["GP"]

df["TS_PCT"] = df["PTS"] / (
    2 * (df["FGA"] + 0.44 * df["FTA"])
)

df["EFG_PCT"] = (
    df["FGM"] + 0.5 * df["FG3M"]
) / df["FGA"]

df["THREE_PA_RATE"] = df["FG3A"] / df["FGA"]

df["FTA_RATE"] = df["FTA"] / df["FGA"]
```

```python
# Kobe career reference values
total_gp = df["GP"].sum()

kobe_reference = {
    "MPG": df["MIN"].sum() / total_gp,
    "PPG": df["PTS"].sum() / total_gp,
    "APG": df["AST"].sum() / total_gp,
    "RPG": df["REB"].sum() / total_gp,
    "SPG": df["STL"].sum() / total_gp,
    "BPG": df["BLK"].sum() / total_gp,
    "TOPG": df["TOV"].sum() / total_gp,

    "FG_PCT": df["FGM"].sum() / df["FGA"].sum(),
    "FG3_PCT": df["FG3M"].sum() / df["FG3A"].sum(),
    "FT_PCT": df["FTM"].sum() / df["FTA"].sum(),

    "TS_PCT": df["PTS"].sum() / (
        2 * (df["FGA"].sum() + 0.44 * df["FTA"].sum())
    ),

    "EFG_PCT": (
        df["FGM"].sum() + 0.5 * df["FG3M"].sum()
    ) / df["FGA"].sum(),

    "THREE_PA_RATE": df["FG3A"].sum() / df["FGA"].sum(),
    "FTA_RATE": df["FTA"].sum() / df["FGA"].sum()
}

kobe_reference
```

```python
def similarity_score(reference, candidate):
    if reference == 0:
        return 100 if candidate == 0 else 0

    difference = abs(candidate - reference)
    percentage_difference = difference / abs(reference)

    return max(0, 100 * (1 - percentage_difference))
```

```python
comparison_metrics = [
    "PTS",
    "AST",
    "REB",
    "STL",
    "BLK",
    "TOV",
    "FG_PCT",
    "FG3_PCT",
    "FT_PCT",
    "TS_PCT",
    "EFG_PCT",
    "THREE_PA_RATE",
    "FTA_RATE"
]

for metric in comparison_metrics:
    if metric in candidate_stats.columns and metric in kobe_reference:
        candidate_stats[f"{metric}_SIM"] = candidate_stats[metric].apply(
            lambda x: similarity_score(kobe_reference[metric], x)
        )

candidate_stats[
    ["PLAYER_NAME"] +
    [f"{m}_SIM" for m in comparison_metrics if f"{m}_SIM" in candidate_stats.columns]
].head()
```

```python
similarity_columns = [
    f"{m}_SIM"
    for m in comparison_metrics
    if f"{m}_SIM" in candidate_stats.columns
]

candidate_stats["TRADITIONAL_SIMILARITY"] = (
    candidate_stats[similarity_columns].mean(axis=1)
)

candidate_stats[
    ["PLAYER_NAME", "TRADITIONAL_SIMILARITY"]
].sort_values(
    "TRADITIONAL_SIMILARITY",
    ascending=False
).head(10)
```
